```python
"""
BFS Path Finding - based on the uploaded handwritten Transition Table
Initial State : Library
Goal State    : Hall
(No imports - plain Python list used as the queue)
"""

# ---------------------------------------------------------
# 1) GRAPH (built from the transition table in the notebook)
# ---------------------------------------------------------
graph = {
    "FF": ["C1", "C2", "C3"],

    "C1": ["FF", "TR", "Lab3", "ER1", "Library", "C2"],
    "C2": ["FF", "HOD", "R13", "R12", "R11", "FW", "SR", "FR",
           "MUR", "MShR", "MNaR", "C1"],
    "C3": ["FF", "Lab4", "ER2", "ER1", "St1"],

    "ER1": ["Library", "C1", "C3"],
    "ER2": ["C3", "Library", "Lab4"],
    "Library": ["ER2", "ER1", "C1"],

    "Lab4": ["ER2", "C3"],
    "St1": ["St2", "C3"],
    "St2": ["St1", "C4"],

    "GF": ["C4", "C5", "C6"],

    "C4": ["GF", "Lab2", "AdOff1", "Hall", "St2"],
    "C5": ["GF", "SSalimR", "R3", "R2", "R1", "BW", "AdOff2",
           "SKhR", "SAkhR", "GCR", "SirKashR", "MSalR", "C6"],
    "C6": ["GF", "SFarR", "Lab1", "ER3", "Hall", "C5"],

    "Hall": ["C4", "C6", "AdOff1"],

    # dead-end rooms (each connects only to its own corridor)
    "HOD": ["C2"], "R13": ["C2"], "R12": ["C2"], "R11": ["C2"],
    "FW": ["C2"], "SR": ["C2"], "FR": ["C2"], "MUR": ["C2"],
    "MShR": ["C2"], "MNaR": ["C2"],
    "TR": ["C1"], "Lab3": ["C1"],

    "Lab2": ["C4"], "AdOff1": ["C4", "Hall"], "AdOff2": ["C5"],
    "SSalimR": ["C5"], "R3": ["C5"], "R2": ["C5"], "R1": ["C5"],
    "BW": ["C5"], "SKhR": ["C5"], "SAkhR": ["C5"], "GCR": ["C5"],
    "SirKashR": ["C5"], "MSalR": ["C5"],
    "SFarR": ["C6"], "Lab1": ["C6"], "ER3": ["C6"],
}


# ---------------------------------------------------------
# 2) BFS FUNCTION (with inqueue / dequeue trace) - no imports
# ---------------------------------------------------------
def bfs(graph, start, goal):
    visited = [start]        # plain list
    parent = {start: None}
    queue = [start]           # plain list used as the queue

    step = 1
    print(f"{'Step':<5}{'Dequeue':<12}{'Inqueue (newly added)':<55}{'Queue after step'}")
    print("-" * 110)

    while len(queue) > 0:
        node = queue.pop(0)      # DEQUEUE (front of list)
        newly_added = []

        if node == goal:
            print(f"{step:<5}{node:<12}{'--- GOAL FOUND ---':<55}{queue}")
            break

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.append(neighbor)
                parent[neighbor] = node
                queue.append(neighbor)   # ENQUEUE (back of list)
                newly_added.append(neighbor)

        print(f"{step:<5}{node:<12}{', '.join(newly_added) if newly_added else '--':<55}{queue}")
        step += 1

    return parent


# ---------------------------------------------------------
# 3) PATH RECONSTRUCTION
# ---------------------------------------------------------
def reconstruct_path(parent, start, goal):
    if goal not in parent:
        return None
    path = []
    node = goal
    while node is not None:
        path.append(node)
        node = parent[node]
    path.reverse()
    return path


# ---------------------------------------------------------
# 4) RUN
# ---------------------------------------------------------
if __name__ == "__main__":
    start_state = "Library"
    goal_state = "Hall"

    parent_map = bfs(graph, start_state, goal_state)
    shortest_path = reconstruct_path(parent_map, start_state, goal_state)

    print("\n--- RESULT ---")
    if shortest_path:
        print("Shortest Path :", " -> ".join(shortest_path))
        print("Total Moves   :", len(shortest_path) - 1)
        print("Total Nodes   :", len(shortest_path))
    else:
        print(f"No path found between {start_state} and {goal_state}.")
```
